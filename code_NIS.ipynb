```python
# ============================================================

#
# Dataset:
#   NSL-KDD-derived network intrusion dataset
#
# Training:
#   Train_data.csv
#   125,973 samples
#
# Independent Test:
#   test_data.csv
#   10,000 samples
#
# Target:
#   xAttack
#
# Classes:
#   normal
#   dos
#   probe
#   r2l
#   u2r
#
# 
# ------------------------------------------------------------
# 1. Independent predefined test set
# 2. Internal validation split from training data
# 3. No test-set leakage
# 4. Automatic removal of Unnamed index columns
# 5. Robust categorical encoding
# 6. Constant-feature removal
# 7. Correlation filtering
# 8. Training-only feature selection
# 9. Class imbalance handling
# 10. Stratified 5-fold cross-validation
# 11. Multiple baseline classifiers
# 12. Hybrid probability-level ensemble
# 13. Macro-F1 / balanced accuracy emphasis
# 14. Per-class precision/recall/F1
# 15. False-negative-rate analysis
# 16. Normal-vs-attack evaluation
# 17. ROC-AUC
# 18. Precision-Recall AUC
# 19. Confusion matrices
# 20. Feature importance
# 21. Model comparison
# 22. Reproducibility artifacts
# 23. Final independent test evaluation
# ============================================================


# ============================================================
# 0. INSTALLATION
# ============================================================

# If required:
#
# !pip install pandas numpy scikit-learn xgboost matplotlib seaborn joblib openpyxl


# ============================================================
# 1. IMPORTS
# ============================================================

import os
import gc
import random
import warnings
import joblib

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import (
    OrdinalEncoder,
    LabelEncoder,
    StandardScaler
)

from sklearn.compose import ColumnTransformer

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.feature_selection import (
    VarianceThreshold
)

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix,
    matthews_corrcoef,
    roc_curve,
    precision_recall_curve
)

from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier
)

from sklearn.linear_model import (
    LogisticRegression
)

from sklearn.neighbors import (
    KNeighborsClassifier
)

from sklearn.naive_bayes import (
    GaussianNB
)

from xgboost import XGBClassifier


warnings.filterwarnings("ignore")


# ============================================================
# 2. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

os.environ["PYTHONHASHSEED"] = str(SEED)


# ============================================================
# 3. CONFIGURATION
# ============================================================

TRAIN_PATH = (
    r"E:\2026مشاريع الملك سطام\Enhancing Sustainable Cybersecurity through Advanced"
    r"\Paper 2\Network-Intrusion-Detection-ML-Project"
    r"\Network-Intrusion-Detection-ML-Project-master"
    r"\Train_data.csv"
)

TEST_PATH = (
    r"E:\2026مشاريع الملك سطام\Enhancing Sustainable Cybersecurity through Advanced"
    r"\Paper 2\Network-Intrusion-Detection-ML-Project"
    r"\Network-Intrusion-Detection-ML-Project-master"
    r"\test_data.csv"
)


TARGET_COLUMN = "xAttack"

VALIDATION_SIZE = 0.20

CORRELATION_THRESHOLD = 0.90

TOP_K_FEATURES = 26

N_CV_FOLDS = 5

OUTPUT_DIR = "NSL_KDD_Results"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 4. DISPLAY CONFIGURATION
# ============================================================

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.width",
    200
)

pd.set_option(
    "display.max_rows",
    200
)


# ============================================================
# 5. LOAD TRAINING DATA
# ============================================================

print("=" * 110)
print("LOADING TRAINING DATA")
print("=" * 110)


train_df = pd.read_csv(
    TRAIN_PATH
)


print(
    "\nTraining dataset loaded successfully."
)

print(
    "Training shape:",
    train_df.shape
)

display(
    train_df.head()
)


# ============================================================
# 6. LOAD INDEPENDENT TEST DATA
# ============================================================

print("\n" + "=" * 110)
print("LOADING INDEPENDENT TEST DATA")
print("=" * 110)


test_df = pd.read_csv(
    TEST_PATH
)


print(
    "\nTest dataset loaded successfully."
)

print(
    "Test shape:",
    test_df.shape
)

display(
    test_df.head()
)


# ============================================================
# 7. REMOVE AUTOMATIC INDEX COLUMNS
# ============================================================

print("\n" + "=" * 110)
print("REMOVING NON-INFORMATIVE INDEX COLUMNS")
print("=" * 110)


def remove_index_columns(df):

    df = df.copy()

    columns_to_remove = []

    for column in df.columns:

        column_lower = str(column).lower()

        if (
            column_lower.startswith("unnamed:")
            or column_lower in [
                "index",
                "id"
            ]
        ):

            columns_to_remove.append(
                column
            )

    if columns_to_remove:

        print(
            "\nRemoving:",
            columns_to_remove
        )

        df = df.drop(
            columns=columns_to_remove
        )

    return df


train_df = remove_index_columns(
    train_df
)

test_df = remove_index_columns(
    test_df
)


print(
    "\nTraining shape after cleanup:",
    train_df.shape
)

print(
    "Test shape after cleanup:",
    test_df.shape
)


# ============================================================
# 8. DATA VALIDATION
# ============================================================

print("\n" + "=" * 110)
print("DATA VALIDATION")
print("=" * 110)


if TARGET_COLUMN not in train_df.columns:

    raise ValueError(
        f"Target '{TARGET_COLUMN}' not found in training data."
    )


if TARGET_COLUMN not in test_df.columns:

    raise ValueError(
        f"Target '{TARGET_COLUMN}' not found in test data."
    )


train_features = [
    c for c in train_df.columns
    if c != TARGET_COLUMN
]

test_features = [
    c for c in test_df.columns
    if c != TARGET_COLUMN
]


missing_in_test = set(
    train_features
) - set(
    test_features
)


extra_in_test = set(
    test_features
) - set(
    train_features
)


print(
    "\nFeatures missing in test:",
    missing_in_test
)

print(
    "Additional features in test:",
    extra_in_test
)


if missing_in_test:

    raise ValueError(
        "Test dataset is missing training features."
    )


# Extra test-only columns are removed

if extra_in_test:

    print(
        "\nRemoving test-only columns:",
        extra_in_test
    )

    test_df = test_df.drop(
        columns=list(extra_in_test)
    )


# Reorder test columns exactly like training

test_df = test_df[
    train_features + [TARGET_COLUMN]
]


print(
    "\n✓ Training and test feature schemas are aligned."
)

print(
    "Number of input features:",
    len(train_features)
)


# ============================================================
# 9. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 110)
print("DUPLICATE ANALYSIS")
print("=" * 110)


train_duplicates = train_df.duplicated().sum()

test_duplicates = test_df.duplicated().sum()


print(
    "\nTraining duplicate rows:",
    train_duplicates
)

print(
    "Test duplicate rows:",
    test_duplicates
)


# ------------------------------------------------------------
# IMPORTANT:
# Do NOT automatically remove duplicates from the test set.
# The test set must remain unchanged for independent evaluation.
# ------------------------------------------------------------


# ============================================================
# 10. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 110)
print("MISSING VALUE ANALYSIS")
print("=" * 110)


train_missing = train_df.isnull().sum().sum()

test_missing = test_df.isnull().sum().sum()


print(
    "\nTraining missing values:",
    train_missing
)

print(
    "Test missing values:",
    test_missing
)


# ============================================================
# 11. TARGET DISTRIBUTION
# ============================================================

print("\n" + "=" * 110)
print("TARGET CLASS DISTRIBUTION")
print("=" * 110)


train_distribution = (

    train_df[TARGET_COLUMN]
    .value_counts()
    .rename_axis("Class")
    .reset_index(name="Training_Instances")
)


test_distribution = (

    test_df[TARGET_COLUMN]
    .value_counts()
    .rename_axis("Class")
    .reset_index(name="Test_Instances")
)


class_distribution = (

    train_distribution
    .merge(
        test_distribution,
        on="Class",
        how="outer"
    )
    .fillna(0)

)


class_distribution["Training_Percentage"] = (

    class_distribution["Training_Instances"]
    / len(train_df)
    * 100

)


class_distribution["Test_Percentage"] = (

    class_distribution["Test_Instances"]
    / len(test_df)
    * 100

)


display(
    class_distribution
)


class_distribution.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Class_Distribution.csv"
    ),

    index=False

)


# ============================================================
# 12. CLASS DISTRIBUTION VISUALIZATION
# ============================================================

plt.figure(
    figsize=(11, 7)
)


x = np.arange(
    len(class_distribution)
)


width = 0.35


plt.bar(
    x - width / 2,
    class_distribution["Training_Instances"],
    width,
    label="Training"
)


plt.bar(
    x + width / 2,
    class_distribution["Test_Instances"],
    width,
    label="Independent Test"
)


plt.xticks(
    x,
    class_distribution["Class"]
)


plt.xlabel(
    "Attack Class"
)


plt.ylabel(
    "Number of Instances"
)


plt.title(
    "Training and Independent Test Class Distribution"
)


plt.legend()


plt.tight_layout()


plt.savefig(
    os.path.join(
        OUTPUT_DIR,
        "Class_Distribution_Comparison.png"
    ),
    dpi=600,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 13. TRAIN / INTERNAL VALIDATION SPLIT
# ============================================================

print("\n" + "=" * 110)
print("TRAINING / INTERNAL VALIDATION SPLIT")
print("=" * 110)


X_full = train_df.drop(
    columns=[TARGET_COLUMN]
).copy()


y_full = train_df[
    TARGET_COLUMN
].astype(str).copy()


X_test_raw = test_df.drop(
    columns=[TARGET_COLUMN]
).copy()


y_test_raw = test_df[
    TARGET_COLUMN
].astype(str).copy()


X_train_raw, X_val_raw, y_train_raw, y_val_raw = train_test_split(

    X_full,

    y_full,

    test_size=VALIDATION_SIZE,

    random_state=SEED,

    stratify=y_full

)


print(
    "\nDevelopment training samples:",
    len(X_train_raw)
)

print(
    "Internal validation samples:",
    len(X_val_raw)
)

print(
    "Independent test samples:",
    len(X_test_raw)
)


# ============================================================
# 14. TARGET ENCODING
# ============================================================

print("\n" + "=" * 110)
print("TARGET ENCODING")
print("=" * 110)


target_encoder = LabelEncoder()


y_train = target_encoder.fit_transform(
    y_train_raw
)


y_val = target_encoder.transform(
    y_val_raw
)


y_test = target_encoder.transform(
    y_test_raw
)


CLASS_NAMES = list(
    target_encoder.classes_
)


NUM_CLASSES = len(
    CLASS_NAMES
)


for i, class_name in enumerate(
    CLASS_NAMES
):

    print(
        f"{i}: {class_name}"
    )


# ============================================================
# 15. FEATURE TYPE IDENTIFICATION
# ============================================================

categorical_columns = (

    X_train_raw
    .select_dtypes(
        include=["object", "category"]
    )
    .columns
    .tolist()

)


numeric_columns = (

    X_train_raw
    .select_dtypes(
        include=[np.number]
    )
    .columns
    .tolist()

)


print("\n" + "=" * 110)
print("FEATURE TYPES")
print("=" * 110)


print(
    "\nCategorical features:"
)

print(
    categorical_columns
)


print(
    "\nNumber of categorical features:",
    len(categorical_columns)
)


print(
    "\nNumber of numeric features:",
    len(numeric_columns)
)


# ============================================================
# 16. ROBUST CATEGORICAL ENCODING
# ============================================================

print("\n" + "=" * 110)
print("CATEGORICAL ENCODING")
print("=" * 110)


# ------------------------------------------------------------
# OrdinalEncoder is used instead of LabelEncoder for feature
# columns because it can safely handle unseen categories.
# ------------------------------------------------------------

encoder = OrdinalEncoder(

    handle_unknown="use_encoded_value",

    unknown_value=-1

)


X_train_processed = X_train_raw.copy()

X_val_processed = X_val_raw.copy()

X_test_processed = X_test_raw.copy()


if categorical_columns:

    X_train_processed[
        categorical_columns
    ] = encoder.fit_transform(

        X_train_raw[
            categorical_columns
        ].astype(str)

    )


    X_val_processed[
        categorical_columns
    ] = encoder.transform(

        X_val_raw[
            categorical_columns
        ].astype(str)

    )


    X_test_processed[
        categorical_columns
    ] = encoder.transform(

        X_test_raw[
            categorical_columns
        ].astype(str)

    )


print(
    "\n✓ Categorical encoder fitted ONLY on training data."
)


# ============================================================
# 17. NUMERIC CONVERSION
# ============================================================

for column in X_train_processed.columns:

    X_train_processed[column] = pd.to_numeric(

        X_train_processed[column],

        errors="coerce"

    )


    X_val_processed[column] = pd.to_numeric(

        X_val_processed[column],

        errors="coerce"

    )


    X_test_processed[column] = pd.to_numeric(

        X_test_processed[column],

        errors="coerce"

    )


# ============================================================
# 18. TRAINING-ONLY MEDIAN IMPUTATION
# ============================================================

train_medians = (

    X_train_processed
    .median(numeric_only=True)

)


X_train_processed = (

    X_train_processed
    .fillna(train_medians)

)


X_val_processed = (

    X_val_processed
    .fillna(train_medians)

)


X_test_processed = (

    X_test_processed
    .fillna(train_medians)

)


print(
    "\n✓ Missing values handled using training-set medians."
)


# ============================================================
# 19. REMOVE CONSTANT FEATURES
# ============================================================

print("\n" + "=" * 110)
print("CONSTANT FEATURE REMOVAL")
print("=" * 110)


variance_selector = VarianceThreshold(
    threshold=0.0
)


X_train_var = variance_selector.fit_transform(
    X_train_processed
)


X_val_var = variance_selector.transform(
    X_val_processed
)


X_test_var = variance_selector.transform(
    X_test_processed
)


variance_features = (

    X_train_processed.columns[
        variance_selector.get_support()
    ]
    .tolist()

)


removed_constant_features = [

    c

    for c in X_train_processed.columns

    if c not in variance_features

]


print(
    "\nOriginal feature count:",
    X_train_processed.shape[1]
)


print(
    "Remaining feature count:",
    len(variance_features)
)


print(
    "Removed constant features:",
    len(removed_constant_features)
)


# Convert to DataFrame

X_train_var = pd.DataFrame(

    X_train_var,

    columns=variance_features

)


X_val_var = pd.DataFrame(

    X_val_var,

    columns=variance_features

)


X_test_var = pd.DataFrame(

    X_test_var,

    columns=variance_features

)


# ============================================================
# 20. STANDARDIZATION
# ============================================================

print("\n" + "=" * 110)
print("FEATURE STANDARDIZATION")
print("=" * 110)


scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train_var
)


X_val_scaled = scaler.transform(
    X_val_var
)


X_test_scaled = scaler.transform(
    X_test_var
)


X_train_scaled = pd.DataFrame(

    X_train_scaled,

    columns=variance_features

)


X_val_scaled = pd.DataFrame(

    X_val_scaled,

    columns=variance_features

)


X_test_scaled = pd.DataFrame(

    X_test_scaled,

    columns=variance_features

)


print(
    "\n✓ StandardScaler fitted ONLY on development training data."
)


# ============================================================
# 21. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 110)
print("CORRELATION ANALYSIS")
print("=" * 110)


correlation_matrix = (

    X_train_scaled
    .corr()
    .abs()

)


upper_triangle = (

    correlation_matrix.where(

        np.triu(

            np.ones(
                correlation_matrix.shape
            ),

            k=1

        ).astype(bool)

    )

)


correlated_features = [

    column

    for column in upper_triangle.columns

    if any(

        upper_triangle[column]
        > CORRELATION_THRESHOLD

    )

]


print(
    "\nCorrelation threshold:",
    CORRELATION_THRESHOLD
)


print(
    "Highly correlated features:",
    len(correlated_features)
)


print(
    "\nRemoved correlated features:"
)


print(
    correlated_features
)


X_train_corr = X_train_scaled.drop(

    columns=correlated_features

)


X_val_corr = X_val_scaled.drop(

    columns=correlated_features

)


X_test_corr = X_test_scaled.drop(

    columns=correlated_features

)


print(
    "\nFeatures after correlation filtering:",
    X_train_corr.shape[1]
)


# ============================================================
# 22. CORRELATION HEATMAP
# ============================================================

plt.figure(
    figsize=(18, 14)
)


sns.heatmap(

    correlation_matrix,

    cmap="coolwarm",

    center=0,

    linewidths=0.1

)


plt.title(
    "Training-Set Feature Correlation Matrix"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Feature_Correlation_Matrix.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 23. TRAINING-ONLY FEATURE SELECTION
# ============================================================

print("\n" + "=" * 110)
print("TRAINING-ONLY FEATURE SELECTION")
print("=" * 110)


feature_selector_model = RandomForestClassifier(

    n_estimators=300,

    random_state=SEED,

    n_jobs=-1,

    class_weight="balanced_subsample"

)


feature_selector_model.fit(

    X_train_corr,

    y_train

)


feature_importance = pd.DataFrame({

    "Feature":
        X_train_corr.columns,

    "Importance":
        feature_selector_model.feature_importances_

})


feature_importance = (

    feature_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)

)


display(
    feature_importance
)


# ============================================================
# 24. SELECT TOP FEATURES
# ============================================================

if (

    TOP_K_FEATURES is not None

    and

    TOP_K_FEATURES < len(
        feature_importance
    )

):

    selected_features = (

        feature_importance
        .head(TOP_K_FEATURES)
        ["Feature"]
        .tolist()

    )

else:

    selected_features = (

        feature_importance[
            "Feature"
        ]
        .tolist()

    )


X_train_selected = X_train_corr[
    selected_features
].copy()


X_val_selected = X_val_corr[
    selected_features
].copy()


X_test_selected = X_test_corr[
    selected_features
].copy()


print(
    "\nNumber of selected features:",
    len(selected_features)
)


print(
    "\nSelected features:"
)


for i, feature in enumerate(
    selected_features,
    1
):

    print(
        f"{i:02d}. {feature}"
    )


feature_importance.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Training_Feature_Importance.csv"
    ),

    index=False

)


# ============================================================
# 25. FEATURE IMPORTANCE VISUALIZATION
# ============================================================

top_plot = (

    feature_importance
    .head(
        min(
            TOP_K_FEATURES,
            len(feature_importance)
        )
    )
    .sort_values(
        "Importance"
    )

)


plt.figure(
    figsize=(12, 9)
)


plt.barh(

    top_plot["Feature"],

    top_plot["Importance"]

)


plt.xlabel(
    "Random Forest Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Training-Only Feature Selection"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Selected_Feature_Importance.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 26. CLASS WEIGHTS
# ============================================================

print("\n" + "=" * 110)
print("CLASS IMBALANCE ANALYSIS")
print("=" * 110)


class_counts = (

    pd.Series(y_train)
    .value_counts()
    .sort_index()

)


total_training = len(
    y_train
)


CLASS_WEIGHTS = {

    class_id:
        total_training /
        (
            NUM_CLASSES *
            class_counts[class_id]
        )

    for class_id in class_counts.index

}


for class_id, weight in CLASS_WEIGHTS.items():

    print(

        f"{CLASS_NAMES[class_id]:10s} : "
        f"{weight:.6f}"

    )


class_weights_by_name = {

    CLASS_NAMES[k]:
        float(v)

    for k, v in CLASS_WEIGHTS.items()

}


pd.DataFrame({

    "Class":
        CLASS_NAMES,

    "Training_Count":
        [
            class_counts[i]
            for i in range(NUM_CLASSES)
        ],

    "Class_Weight":
        [
            CLASS_WEIGHTS[i]
            for i in range(NUM_CLASSES)
        ]

}).to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Class_Weights.csv"
    ),

    index=False

)


# ============================================================
# 27. SAMPLE WEIGHTS
# ============================================================

sample_weights = np.array([

    CLASS_WEIGHTS[
        int(label)
    ]

    for label in y_train

])


# ============================================================
# 28. DEFINE BASELINE MODELS
# ============================================================

print("\n" + "=" * 110)
print("MODEL CONFIGURATION")
print("=" * 110)


models = {

    "Logistic Regression":

        LogisticRegression(

            max_iter=3000,

            class_weight="balanced",

            random_state=SEED,

            solver="lbfgs",

            multi_class="auto"

        ),


    "Decision Tree":

        __import__(
            "sklearn.tree",
            fromlist=[
                "DecisionTreeClassifier"
            ]
        ).DecisionTreeClassifier(

            random_state=SEED,

            class_weight="balanced",

            max_depth=None,

            min_samples_leaf=2

        ),


    "Random Forest":

        RandomForestClassifier(

            n_estimators=400,

            random_state=SEED,

            class_weight="balanced_subsample",

            n_jobs=-1,

            max_features="sqrt",

            min_samples_leaf=1

        ),


    "Extra Trees":

        ExtraTreesClassifier(

            n_estimators=400,

            random_state=SEED,

            class_weight="balanced",

            n_jobs=-1,

            max_features="sqrt",

            min_samples_leaf=1

        ),


    "KNN":

        KNeighborsClassifier(

            n_neighbors=7,

            weights="distance",

            n_jobs=-1

        ),


    "Gaussian NB":

        GaussianNB(),


    "XGBoost":

        XGBClassifier(

            n_estimators=400,

            max_depth=7,

            learning_rate=0.05,

            subsample=0.85,

            colsample_bytree=0.85,

            objective="multi:softprob",

            num_class=NUM_CLASSES,

            eval_metric="mlogloss",

            random_state=SEED,

            n_jobs=-1,

            tree_method="hist",

            reg_alpha=0.1,

            reg_lambda=1.0

        )

}


for name in models:

    print(
        "✓",
        name
    )


# ============================================================
# 29. MODEL TRAINING
# ============================================================

trained_models = {}

validation_predictions = {}

validation_probabilities = {}

validation_results = {}


for model_name, model in models.items():

    print("\n" + "=" * 110)

    print(
        f"TRAINING: {model_name}"
    )

    print("=" * 110)


    # --------------------------------------------------------
    # Models supporting class/sample weights
    # --------------------------------------------------------

    if model_name in [

        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "Extra Trees"

    ]:

        model.fit(

            X_train_selected,

            y_train

        )


    elif model_name == "XGBoost":

        model.fit(

            X_train_selected,

            y_train,

            sample_weight=sample_weights

        )


    else:

        # KNN and GaussianNB
        # are trained on the original distribution
        model.fit(

            X_train_selected,

            y_train

        )


    trained_models[
        model_name
    ] = model


    # --------------------------------------------------------
    # Validation prediction
    # --------------------------------------------------------

    y_val_pred = model.predict(

        X_val_selected

    )


    y_val_prob = model.predict_proba(

        X_val_selected

    )


    validation_predictions[
        model_name
    ] = y_val_pred


    validation_probabilities[
        model_name
    ] = y_val_prob


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(

        y_val,

        y_val_pred

    )


    balanced_acc = balanced_accuracy_score(

        y_val,

        y_val_pred

    )


    precision_macro = precision_score(

        y_val,

        y_val_pred,

        average="macro",

        zero_division=0

    )


    recall_macro = recall_score(

        y_val,

        y_val_pred,

        average="macro",

        zero_division=0

    )


    f1_macro = f1_score(

        y_val,

        y_val_pred,

        average="macro",

        zero_division=0

    )


    f1_weighted = f1_score(

        y_val,

        y_val_pred,

        average="weighted",

        zero_division=0

    )


    mcc = matthews_corrcoef(

        y_val,

        y_val_pred

    )


    auc = roc_auc_score(

        y_val,

        y_val_prob,

        multi_class="ovr",

        average="macro"

    )


    pr_auc = average_precision_score(

        pd.get_dummies(
            y_val
        ),

        y_val_prob,

        average="macro"

    )


    validation_results[
        model_name
    ] = {

        "Accuracy":
            accuracy,

        "Balanced_Accuracy":
            balanced_acc,

        "Macro_Precision":
            precision_macro,

        "Macro_Recall":
            recall_macro,

        "Macro_F1":
            f1_macro,

        "Weighted_F1":
            f1_weighted,

        "ROC_AUC":
            auc,

        "PR_AUC":
            pr_auc,

        "MCC":
            mcc

    }


    print(
        f"Accuracy          : {accuracy:.4f}"
    )

    print(
        f"Balanced Accuracy : {balanced_acc:.4f}"
    )

    print(
        f"Macro Precision   : {precision_macro:.4f}"
    )

    print(
        f"Macro Recall      : {recall_macro:.4f}"
    )

    print(
        f"Macro F1          : {f1_macro:.4f}"
    )

    print(
        f"Weighted F1       : {f1_weighted:.4f}"
    )

    print(
        f"ROC-AUC           : {auc:.4f}"
    )

    print(
        f"PR-AUC             : {pr_auc:.4f}"
    )

    print(
        f"MCC               : {mcc:.4f}"
    )


# ============================================================
# 30. VALIDATION MODEL COMPARISON
# ============================================================

validation_results_df = (

    pd.DataFrame(
        validation_results
    )
    .T
    .sort_values(
        "Macro_F1",
        ascending=False
    )

)


print("\n" + "=" * 110)
print("INTERNAL VALIDATION MODEL COMPARISON")
print("=" * 110)


display(
    validation_results_df.style.format(
        "{:.4f}"
    )
)


validation_results_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Internal_Validation_Model_Comparison.csv"
    )

)


# ============================================================
# 31. FIVE-FOLD CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 110)
print("5-FOLD STRATIFIED CROSS-VALIDATION")
print("=" * 110)


cv = StratifiedKFold(

    n_splits=N_CV_FOLDS,

    shuffle=True,

    random_state=SEED

)


cv_scoring = {

    "accuracy":
        "accuracy",

    "balanced_accuracy":
        "balanced_accuracy",

    "precision_macro":
        "precision_macro",

    "recall_macro":
        "recall_macro",

    "f1_macro":
        "f1_macro",

    "f1_weighted":
        "f1_weighted"

}


cv_results = {}


for model_name, model in models.items():

    print(
        f"\nCross-validating: {model_name}"
    )


    try:

        scores = cross_validate(

            model,

            X_train_selected,

            y_train,

            cv=cv,

            scoring=cv_scoring,

            n_jobs=-1,

            return_train_score=False

        )


        cv_results[
            model_name
        ] = {

            "CV_Accuracy_Mean":
                scores[
                    "test_accuracy"
                ].mean(),

            "CV_Accuracy_STD":
                scores[
                    "test_accuracy"
                ].std(),

            "CV_Balanced_Accuracy_Mean":
                scores[
                    "test_balanced_accuracy"
                ].mean(),

            "CV_Balanced_Accuracy_STD":
                scores[
                    "test_balanced_accuracy"
                ].std(),

            "CV_Macro_F1_Mean":
                scores[
                    "test_f1_macro"
                ].mean(),

            "CV_Macro_F1_STD":
                scores[
                    "test_f1_macro"
                ].std(),

            "CV_Macro_Recall_Mean":
                scores[
                    "test_recall_macro"
                ].mean(),

            "CV_Macro_Recall_STD":
                scores[
                    "test_recall_macro"
                ].std()

        }


    except Exception as e:

        print(
            "Cross-validation failed:",
            e
        )


cv_results_df = (

    pd.DataFrame(
        cv_results
    )
    .T
    .sort_values(
        "CV_Macro_F1_Mean",
        ascending=False
    )

)


display(
    cv_results_df.style.format(
        "{:.4f}"
    )
)


cv_results_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Five_Fold_Cross_Validation.csv"
    )

)


# ============================================================
# 32. SELECT BEST BASE MODEL
# ============================================================

best_base_model_name = (

    validation_results_df
    ["Macro_F1"]
    .idxmax()

)


print(
    "\nBest baseline according to internal validation Macro-F1:"
)


print(
    best_base_model_name
)


# ============================================================
# 33. HYBRID PROBABILITY-LEVEL ENSEMBLE
# ============================================================

print("\n" + "=" * 110)
print("HYBRID PROBABILITY-LEVEL ENSEMBLE")
print("=" * 110)


# ------------------------------------------------------------
# The hybrid framework combines complementary classifiers.
#
# We do NOT claim that this is a novel algorithm.
#
# The hybrid component consists of:
#
#   Random Forest
#   Extra Trees
#   XGBoost
#   Logistic Regression
#
# Their validation probabilities are combined using weights
# determined from validation Macro-F1.
#
# 
# reads primarily as a model comparison study.
# ------------------------------------------------------------


ensemble_candidates = [

    "Random Forest",
    "Extra Trees",
    "XGBoost",
    "Logistic Regression"

]


ensemble_weights_raw = {}


for name in ensemble_candidates:

    ensemble_weights_raw[
        name
    ] = max(

        validation_results[
            name
        ]["Macro_F1"],

        1e-6

    )


weight_sum = sum(
    ensemble_weights_raw.values()
)


ensemble_weights = {

    name:
        weight / weight_sum

    for name, weight
    in ensemble_weights_raw.items()

}


print(
    "\nHybrid ensemble weights:"
)


for name, weight in ensemble_weights.items():

    print(
        f"{name:25s}: {weight:.4f}"
    )


# ------------------------------------------------------------
# Combine validation probabilities
# ------------------------------------------------------------

ensemble_val_prob = np.zeros(

    (
        len(X_val_selected),
        NUM_CLASSES
    )

)


for name in ensemble_candidates:

    ensemble_val_prob += (

        ensemble_weights[name]
        *
        validation_probabilities[name]

    )


ensemble_val_pred = np.argmax(

    ensemble_val_prob,

    axis=1

)


# ============================================================
# 34. HYBRID VALIDATION METRICS
# ============================================================

ensemble_val_accuracy = accuracy_score(

    y_val,

    ensemble_val_pred

)


ensemble_val_balanced_accuracy = (

    balanced_accuracy_score(

        y_val,

        ensemble_val_pred

    )

)


ensemble_val_macro_precision = (

    precision_score(

        y_val,

        ensemble_val_pred,

        average="macro",

        zero_division=0

    )

)


ensemble_val_macro_recall = (

    recall_score(

        y_val,

        ensemble_val_pred,

        average="macro",

        zero_division=0

    )

)


ensemble_val_macro_f1 = (

    f1_score(

        y_val,

        ensemble_val_pred,

        average="macro",

        zero_division=0

    )

)


ensemble_val_weighted_f1 = (

    f1_score(

        y_val,

        ensemble_val_pred,

        average="weighted",

        zero_division=0

    )

)


ensemble_val_mcc = matthews_corrcoef(

    y_val,

    ensemble_val_pred

)


ensemble_val_auc = roc_auc_score(

    y_val,

    ensemble_val_prob,

    multi_class="ovr",

    average="macro"

)


ensemble_val_pr_auc = average_precision_score(

    pd.get_dummies(
        y_val
    ),

    ensemble_val_prob,

    average="macro"

)


hybrid_validation_results = {

    "Accuracy":
        ensemble_val_accuracy,

    "Balanced_Accuracy":
        ensemble_val_balanced_accuracy,

    "Macro_Precision":
        ensemble_val_macro_precision,

    "Macro_Recall":
        ensemble_val_macro_recall,

    "Macro_F1":
        ensemble_val_macro_f1,

    "Weighted_F1":
        ensemble_val_weighted_f1,

    "ROC_AUC":
        ensemble_val_auc,

    "PR_AUC":
        ensemble_val_pr_auc,

    "MCC":
        ensemble_val_mcc

}


print("\nHybrid validation performance:")


for metric, value in hybrid_validation_results.items():

    print(
        f"{metric:22s}: {value:.4f}"
    )


# ============================================================
# 35. ADD HYBRID TO VALIDATION COMPARISON
# ============================================================

validation_results_df.loc[
    "Hybrid Ensemble"
] = hybrid_validation_results


validation_results_df = (

    validation_results_df
    .sort_values(
        "Macro_F1",
        ascending=False
    )

)


display(
    validation_results_df.style.format(
        "{:.4f}"
    )
)


validation_results_df.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Complete_Validation_Model_Comparison.csv"
    )

)


# ============================================================
# 36. VALIDATION CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 110)
print("HYBRID VALIDATION CLASSIFICATION REPORT")
print("=" * 110)


print(

    classification_report(

        y_val,

        ensemble_val_pred,

        target_names=CLASS_NAMES,

        digits=4,

        zero_division=0

    )

)


# ============================================================
# 37. FALSE NEGATIVE RATE FUNCTION
# ============================================================

def calculate_per_class_fnr(

    y_true,
    y_pred,
    class_names

):

    cm = confusion_matrix(
        y_true,
        y_pred
    )


    rows = []


    for i, class_name in enumerate(
        class_names
    ):

        TP = cm[i, i]

        FN = (
            cm[i, :].sum()
            - TP
        )

        FP = (
            cm[:, i].sum()
            - TP
        )

        TN = (
            cm.sum()
            - TP
            - FN
            - FP
        )


        recall = (

            TP /
            (TP + FN)
            if (TP + FN) > 0
            else 0

        )


        fnr = (

            FN /
            (FN + TP)
            if (FN + TP) > 0
            else 0

        )


        rows.append({

            "Class":
                class_name,

            "TP":
                TP,

            "FN":
                FN,

            "FP":
                FP,

            "TN":
                TN,

            "Recall":
                recall,

            "False_Negative_Rate":
                fnr

        })


    return pd.DataFrame(rows)


# ============================================================
# 38. HYBRID FALSE NEGATIVE ANALYSIS
# ============================================================

print("\n" + "=" * 110)
print("FALSE NEGATIVE ANALYSIS")
print("=" * 110)


validation_fnr = calculate_per_class_fnr(

    y_val,

    ensemble_val_pred,

    CLASS_NAMES

)


display(
    validation_fnr
)


validation_fnr.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Validation_False_Negative_Analysis.csv"
    ),

    index=False

)


# ============================================================
# 39. NORMAL VS ATTACK EVALUATION
# ============================================================

print("\n" + "=" * 110)
print("NORMAL VS ATTACK EVALUATION")
print("=" * 110)


normal_class_id = (

    CLASS_NAMES
    .index("normal")

)


# 1 = attack
# 0 = normal

y_val_binary = (

    y_val != normal_class_id
).astype(int)


y_val_binary_pred = (

    ensemble_val_pred != normal_class_id
).astype(int)


binary_accuracy = accuracy_score(

    y_val_binary,

    y_val_binary_pred

)


binary_precision = precision_score(

    y_val_binary,

    y_val_binary_pred,

    zero_division=0

)


binary_recall = recall_score(

    y_val_binary,

    y_val_binary_pred,

    zero_division=0

)


binary_f1 = f1_score(

    y_val_binary,

    y_val_binary_pred,

    zero_division=0

)


binary_cm = confusion_matrix(

    y_val_binary,

    y_val_binary_pred

)


binary_tn, binary_fp, binary_fn, binary_tp = (

    binary_cm.ravel()

)


binary_fnr = (

    binary_fn /
    (binary_fn + binary_tp)

    if
    (binary_fn + binary_tp) > 0

    else 0

)


print(
    f"Binary Accuracy : {binary_accuracy:.4f}"
)

print(
    f"Attack Precision: {binary_precision:.4f}"
)

print(
    f"Attack Recall   : {binary_recall:.4f}"
)

print(
    f"Attack F1       : {binary_f1:.4f}"
)

print(
    f"Attack FNR      : {binary_fnr:.4f}"
)


binary_results = pd.DataFrame({

    "Metric": [

        "Accuracy",
        "Attack Precision",
        "Attack Recall",
        "Attack F1",
        "Attack FNR"

    ],

    "Value": [

        binary_accuracy,
        binary_precision,
        binary_recall,
        binary_f1,
        binary_fnr

    ]

})


binary_results.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Validation_Normal_vs_Attack.csv"
    ),

    index=False

)


# ============================================================
# 40. HYBRID VALIDATION CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(9, 8)
)


sns.heatmap(

    confusion_matrix(
        y_val,
        ensemble_val_pred
    ),

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=CLASS_NAMES,

    yticklabels=CLASS_NAMES,

    linewidths=0.5

)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "True Class"
)


plt.title(
    "Hybrid Ensemble - Internal Validation Confusion Matrix"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Hybrid_Validation_Confusion_Matrix.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 41. NORMALIZED CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(9, 8)
)


cm_normalized = (

    confusion_matrix(
        y_val,
        ensemble_val_pred,
        normalize="true"
    )

)


sns.heatmap(

    cm_normalized,

    annot=True,

    fmt=".3f",

    cmap="Blues",

    xticklabels=CLASS_NAMES,

    yticklabels=CLASS_NAMES

)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "True Class"
)


plt.title(
    "Hybrid Ensemble - Normalized Confusion Matrix"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Hybrid_Normalized_Confusion_Matrix.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 42. VALIDATION ROC CURVES
# ============================================================

plt.figure(
    figsize=(10, 8)
)


for class_id, class_name in enumerate(
    CLASS_NAMES
):

    y_binary = (

        y_val == class_id
    ).astype(int)


    fpr, tpr, _ = roc_curve(

        y_binary,

        ensemble_val_prob[:, class_id]

    )


    auc_value = roc_auc_score(

        y_binary,

        ensemble_val_prob[:, class_id]

    )


    plt.plot(

        fpr,

        tpr,

        label=f"{class_name} (AUC={auc_value:.4f})"

    )


plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)


plt.xlabel(
    "False Positive Rate"
)


plt.ylabel(
    "True Positive Rate"
)


plt.title(
    "Hybrid Ensemble One-vs-Rest ROC Curves"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Hybrid_Validation_ROC.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 43. VALIDATION PRECISION-RECALL CURVES
# ============================================================

plt.figure(
    figsize=(10, 8)
)


for class_id, class_name in enumerate(
    CLASS_NAMES
):

    y_binary = (

        y_val == class_id
    ).astype(int)


    precision, recall, _ = (

        precision_recall_curve(

            y_binary,

            ensemble_val_prob[
                :,
                class_id
            ]

        )

    )


    ap = average_precision_score(

        y_binary,

        ensemble_val_prob[
            :,
            class_id
        ]

    )


    plt.plot(

        recall,

        precision,

        label=f"{class_name} (AP={ap:.4f})"

    )


plt.xlabel(
    "Recall"
)


plt.ylabel(
    "Precision"
)


plt.title(
    "Hybrid Ensemble Precision-Recall Curves"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Hybrid_Validation_PR_Curves.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 44. MODEL COMPARISON PLOT
# ============================================================

plot_metrics = [

    "Accuracy",
    "Balanced_Accuracy",
    "Macro_F1",
    "Weighted_F1"

]


comparison_plot = (

    validation_results_df[
        plot_metrics
    ]

)


comparison_plot.plot(

    kind="bar",

    figsize=(15, 8)

)


plt.ylabel(
    "Score"
)


plt.xlabel(
    "Model"
)


plt.title(
    "Model Comparison"
)


plt.xticks(
    rotation=30,
    ha="right"
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Model_Comparison.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 45. FINAL MODEL SELECTION
# ============================================================

print("\n" + "=" * 110)
print("FINAL MODEL SELECTION")
print("=" * 110)


# ------------------------------------------------------------
# Primary criterion:
# Macro-F1
#
# Secondary:
# Balanced Accuracy
# ------------------------------------------------------------

best_model_name = (

    validation_results_df
    .sort_values(
        [
            "Macro_F1",
            "Balanced_Accuracy"
        ],
        ascending=False
    )
    .index[0]

)


print(
    "\nSelected final framework:"
)


print(
    best_model_name
)


# ============================================================
# 46. RETRAIN BASE MODELS ON COMPLETE TRAINING DATA
# ============================================================

print("\n" + "=" * 110)
print("FINAL TRAINING ON COMPLETE TRAINING DATA")
print("=" * 110)


# ------------------------------------------------------------
# Important:
#
# The independent test set is NOT used here.
#
# We reconstruct the preprocessing using the complete training
# data so that the final model can use all 125,973 training
# observations.
# ------------------------------------------------------------


X_complete_raw = train_df.drop(

    columns=[TARGET_COLUMN]
).copy()


y_complete_raw = train_df[
    TARGET_COLUMN
].astype(str)


y_complete = target_encoder.transform(

    y_complete_raw

)


# ------------------------------------------------------------
# Categorical encoding
# ------------------------------------------------------------

X_complete_processed = (

    X_complete_raw.copy()

)


if categorical_columns:

    X_complete_processed[
        categorical_columns
    ] = encoder.transform(

        X_complete_raw[
            categorical_columns
        ].astype(str)

    )


# ------------------------------------------------------------
# Numeric conversion
# ------------------------------------------------------------

for column in X_complete_processed.columns:

    X_complete_processed[column] = (

        pd.to_numeric(

            X_complete_processed[column],

            errors="coerce"

        )

    )


# ------------------------------------------------------------
# Use complete-training medians
# ------------------------------------------------------------

complete_medians = (

    X_complete_processed
    .median(
        numeric_only=True
    )

)


X_complete_processed = (

    X_complete_processed
    .fillna(
        complete_medians
    )

)


X_test_final_processed = X_test_raw.copy()


if categorical_columns:

    X_test_final_processed[
        categorical_columns
    ] = encoder.transform(

        X_test_raw[
            categorical_columns
        ].astype(str)

    )


for column in X_test_final_processed.columns:

    X_test_final_processed[column] = (

        pd.to_numeric(

            X_test_final_processed[column],

            errors="coerce"

        )

    )


X_test_final_processed = (

    X_test_final_processed
    .fillna(
        complete_medians
    )

)


# ------------------------------------------------------------
# Variance filtering
# ------------------------------------------------------------

final_variance_selector = VarianceThreshold(
    threshold=0.0
)


X_complete_var = (

    final_variance_selector
    .fit_transform(
        X_complete_processed
    )

)


X_test_final_var = (

    final_variance_selector
    .transform(
        X_test_final_processed
    )

)


final_variance_features = (

    X_complete_processed.columns[
        final_variance_selector.get_support()
    ]
    .tolist()

)


X_complete_var = pd.DataFrame(

    X_complete_var,

    columns=final_variance_features

)


X_test_final_var = pd.DataFrame(

    X_test_final_var,

    columns=final_variance_features

)


# ------------------------------------------------------------
# Scaling
# ------------------------------------------------------------

final_scaler = StandardScaler()


X_complete_scaled = final_scaler.fit_transform(

    X_complete_var

)


X_test_final_scaled = final_scaler.transform(

    X_test_final_var

)


X_complete_scaled = pd.DataFrame(

    X_complete_scaled,

    columns=final_variance_features

)


X_test_final_scaled = pd.DataFrame(

    X_test_final_scaled,

    columns=final_variance_features

)


# ------------------------------------------------------------
# Correlation filtering
# ------------------------------------------------------------

final_corr_matrix = (

    X_complete_scaled
    .corr()
    .abs()

)


final_upper = (

    final_corr_matrix.where(

        np.triu(

            np.ones(
                final_corr_matrix.shape
            ),

            k=1

        ).astype(bool)

    )

)


final_correlated_features = [

    column

    for column in final_upper.columns

    if any(

        final_upper[column]
        > CORRELATION_THRESHOLD

    )

]


X_complete_corr = (

    X_complete_scaled
    .drop(
        columns=final_correlated_features
    )

)


X_test_final_corr = (

    X_test_final_scaled
    .drop(
        columns=final_correlated_features
    )

)


# ------------------------------------------------------------
# Final feature selection
# ------------------------------------------------------------

final_selector_model = RandomForestClassifier(

    n_estimators=300,

    random_state=SEED,

    n_jobs=-1,

    class_weight="balanced_subsample"

)


final_selector_model.fit(

    X_complete_corr,

    y_complete

)


final_importance = pd.DataFrame({

    "Feature":
        X_complete_corr.columns,

    "Importance":
        final_selector_model.feature_importances_

})


final_importance = (

    final_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)

)


final_selected_features = (

    final_importance
    .head(
        min(
            TOP_K_FEATURES,
            len(final_importance)
        )
    )
    ["Feature"]
    .tolist()

)


X_complete_selected = (

    X_complete_corr[
        final_selected_features
    ]

)


X_test_final_selected = (

    X_test_final_corr[
        final_selected_features
    ]

)


print(
    "\nFinal training matrix:",
    X_complete_selected.shape
)


print(
    "Final independent test matrix:",
    X_test_final_selected.shape
)


# ============================================================
# 47. FINAL CLASS WEIGHTS
# ============================================================

complete_class_counts = (

    pd.Series(
        y_complete
    )
    .value_counts()
    .sort_index()

)


complete_class_weights = {

    class_id:

        len(y_complete) /
        (
            NUM_CLASSES *
            complete_class_counts[class_id]
        )

    for class_id
    in complete_class_counts.index

}


complete_sample_weights = np.array([

    complete_class_weights[
        int(label)
    ]

    for label in y_complete

])


# ============================================================
# 48. TRAIN FINAL BASE MODELS
# ============================================================

final_models = {}


final_models["Logistic Regression"] = (

    LogisticRegression(

        max_iter=3000,

        class_weight="balanced",

        random_state=SEED

    )

)


final_models["Random Forest"] = (

    RandomForestClassifier(

        n_estimators=500,

        random_state=SEED,

        class_weight="balanced_subsample",

        n_jobs=-1,

        max_features="sqrt"

    )

)


final_models["Extra Trees"] = (

    ExtraTreesClassifier(

        n_estimators=500,

        random_state=SEED,

        class_weight="balanced",

        n_jobs=-1,

        max_features="sqrt"

    )

)


final_models["XGBoost"] = (

    XGBClassifier(

        n_estimators=500,

        max_depth=7,

        learning_rate=0.05,

        subsample=0.85,

        colsample_bytree=0.85,

        objective="multi:softprob",

        num_class=NUM_CLASSES,

        eval_metric="mlogloss",

        random_state=SEED,

        n_jobs=-1,

        tree_method="hist",

        reg_alpha=0.1,

        reg_lambda=1.0

    )

)


for name, model in final_models.items():

    print(
        f"\nTraining final {name}..."
    )


    if name == "XGBoost":

        model.fit(

            X_complete_selected,

            y_complete,

            sample_weight=complete_sample_weights

        )

    else:

        model.fit(

            X_complete_selected,

            y_complete

        )


# ============================================================
# 49. FINAL HYBRID TEST PREDICTION
# ============================================================

print("\n" + "=" * 110)
print("INDEPENDENT TEST EVALUATION")
print("=" * 110)


test_probabilities = {}


for name, model in final_models.items():

    test_probabilities[name] = (

        model.predict_proba(

            X_test_final_selected

        )

    )


# ------------------------------------------------------------
# Use validation-derived ensemble weights.
# ------------------------------------------------------------

final_ensemble_prob = np.zeros(

    (
        len(X_test_final_selected),
        NUM_CLASSES
    )

)


for name in ensemble_candidates:

    final_ensemble_prob += (

        ensemble_weights[name]
        *
        test_probabilities[name]

    )


final_ensemble_pred = np.argmax(

    final_ensemble_prob,

    axis=1

)


# ============================================================
# 50. FINAL TEST METRICS
# ============================================================

test_accuracy = accuracy_score(

    y_test,

    final_ensemble_pred

)


test_balanced_accuracy = (

    balanced_accuracy_score(

        y_test,

        final_ensemble_pred

    )

)


test_macro_precision = precision_score(

    y_test,

    final_ensemble_pred,

    average="macro",

    zero_division=0

)


test_macro_recall = recall_score(

    y_test,

    final_ensemble_pred,

    average="macro",

    zero_division=0

)


test_macro_f1 = f1_score(

    y_test,

    final_ensemble_pred,

    average="macro",

    zero_division=0

)


test_weighted_f1 = f1_score(

    y_test,

    final_ensemble_pred,

    average="weighted",

    zero_division=0

)


test_mcc = matthews_corrcoef(

    y_test,

    final_ensemble_pred

)


test_auc = roc_auc_score(

    y_test,

    final_ensemble_prob,

    multi_class="ovr",

    average="macro"

)


test_pr_auc = average_precision_score(

    pd.get_dummies(
        y_test
    ),

    final_ensemble_prob,

    average="macro"

)


print(
    "\nFINAL INDEPENDENT TEST PERFORMANCE"
)


print(
    f"\nAccuracy           : {test_accuracy:.4f}"
)

print(
    f"Balanced Accuracy  : {test_balanced_accuracy:.4f}"
)

print(
    f"Macro Precision    : {test_macro_precision:.4f}"
)

print(
    f"Macro Recall       : {test_macro_recall:.4f}"
)

print(
    f"Macro F1           : {test_macro_f1:.4f}"
)

print(
    f"Weighted F1        : {test_weighted_f1:.4f}"
)

print(
    f"ROC-AUC            : {test_auc:.4f}"
)

print(
    f"PR-AUC             : {test_pr_auc:.4f}"
)

print(
    f"MCC                : {test_mcc:.4f}"
)


# ============================================================
# 51. FINAL TEST CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 110)
print("FINAL INDEPENDENT TEST CLASSIFICATION REPORT")
print("=" * 110)


test_report = classification_report(

    y_test,

    final_ensemble_pred,

    target_names=CLASS_NAMES,

    digits=4,

    zero_division=0

)


print(
    test_report
)


with open(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_Classification_Report.txt"
    ),

    "w",

    encoding="utf-8"

) as f:

    f.write(
        test_report
    )


# ============================================================
# 52. FINAL TEST FALSE NEGATIVE ANALYSIS
# ============================================================

test_fnr = calculate_per_class_fnr(

    y_test,

    final_ensemble_pred,

    CLASS_NAMES

)


print("\n" + "=" * 110)
print("FINAL TEST FALSE NEGATIVE ANALYSIS")
print("=" * 110)


display(
    test_fnr
)


test_fnr.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_False_Negative_Analysis.csv"
    ),

    index=False

)


# ============================================================
# 53. FINAL NORMAL VS ATTACK ANALYSIS
# ============================================================

y_test_binary = (

    y_test != normal_class_id
).astype(int)


test_binary_pred = (

    final_ensemble_pred != normal_class_id
).astype(int)


test_binary_accuracy = accuracy_score(

    y_test_binary,

    test_binary_pred

)


test_binary_precision = precision_score(

    y_test_binary,

    test_binary_pred,

    zero_division=0

)


test_binary_recall = recall_score(

    y_test_binary,

    test_binary_pred,

    zero_division=0

)


test_binary_f1 = f1_score(

    y_test_binary,

    test_binary_pred,

    zero_division=0

)


test_binary_cm = confusion_matrix(

    y_test_binary,

    test_binary_pred

)


tn, fp, fn, tp = test_binary_cm.ravel()


test_binary_fnr = (

    fn /
    (fn + tp)

    if
    (fn + tp) > 0

    else 0

)


print("\n" + "=" * 110)
print("FINAL NORMAL VS ATTACK PERFORMANCE")
print("=" * 110)


print(
    f"\nAccuracy       : {test_binary_accuracy:.4f}"
)

print(
    f"Attack Precision: {test_binary_precision:.4f}"
)

print(
    f"Attack Recall   : {test_binary_recall:.4f}"
)

print(
    f"Attack F1       : {test_binary_f1:.4f}"
)

print(
    f"Attack FNR      : {test_binary_fnr:.4f}"
)


# ============================================================
# 54. FINAL TEST CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(9, 8)
)


sns.heatmap(

    confusion_matrix(
        y_test,
        final_ensemble_pred
    ),

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=CLASS_NAMES,

    yticklabels=CLASS_NAMES,

    linewidths=0.5

)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "True Class"
)


plt.title(
    "Hybrid Ensemble - Independent Test Confusion Matrix"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_Confusion_Matrix.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 55. NORMALIZED TEST CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(9, 8)
)


sns.heatmap(

    confusion_matrix(

        y_test,

        final_ensemble_pred,

        normalize="true"

    ),

    annot=True,

    fmt=".3f",

    cmap="Blues",

    xticklabels=CLASS_NAMES,

    yticklabels=CLASS_NAMES

)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "True Class"
)


plt.title(
    "Hybrid Ensemble - Normalized Independent Test Confusion Matrix"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_Normalized_Confusion_Matrix.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 56. FINAL ROC CURVES
# ============================================================

plt.figure(
    figsize=(10, 8)
)


for class_id, class_name in enumerate(
    CLASS_NAMES
):

    y_binary = (

        y_test == class_id
    ).astype(int)


    fpr, tpr, _ = roc_curve(

        y_binary,

        final_ensemble_prob[
            :,
            class_id
        ]

    )


    auc_value = roc_auc_score(

        y_binary,

        final_ensemble_prob[
            :,
            class_id
        ]

    )


    plt.plot(

        fpr,

        tpr,

        label=f"{class_name} (AUC={auc_value:.4f})"

    )


plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--"

)


plt.xlabel(
    "False Positive Rate"
)


plt.ylabel(
    "True Positive Rate"
)


plt.title(
    "Independent Test One-vs-Rest ROC Curves"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_ROC_Curves.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 57. FINAL TEST PRECISION-RECALL CURVES
# ============================================================

plt.figure(
    figsize=(10, 8)
)


for class_id, class_name in enumerate(
    CLASS_NAMES
):

    y_binary = (

        y_test == class_id
    ).astype(int)


    precision, recall, _ = (

        precision_recall_curve(

            y_binary,

            final_ensemble_prob[
                :,
                class_id
            ]

        )

    )


    ap = average_precision_score(

        y_binary,

        final_ensemble_prob[
            :,
            class_id
        ]

    )


    plt.plot(

        recall,

        precision,

        label=f"{class_name} (AP={ap:.4f})"

    )


plt.xlabel(
    "Recall"
)


plt.ylabel(
    "Precision"
)


plt.title(
    "Independent Test Precision-Recall Curves"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Final_Test_PR_Curves.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 58. XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_final_model = final_models[
    "XGBoost"
]


xgb_importance = pd.DataFrame({

    "Feature":
        final_selected_features,

    "Importance":
        xgb_final_model.feature_importances_

})


xgb_importance = (

    xgb_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .reset_index(drop=True)

)


display(
    xgb_importance
)


xgb_importance.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Final_XGBoost_Feature_Importance.csv"
    ),

    index=False

)


plt.figure(
    figsize=(12, 9)
)


plot_xgb = (

    xgb_importance
    .head(
        min(
            20,
            len(xgb_importance)
        )
    )
    .sort_values(
        "Importance"
    )

)


plt.barh(

    plot_xgb["Feature"],

    plot_xgb["Importance"]

)


plt.xlabel(
    "XGBoost Feature Importance"
)


plt.ylabel(
    "Feature"
)


plt.title(
    "Top XGBoost Features"
)


plt.tight_layout()


plt.savefig(

    os.path.join(
        OUTPUT_DIR,
        "Final_XGBoost_Feature_Importance.png"
    ),

    dpi=600,

    bbox_inches="tight"

)


plt.show()


# ============================================================
# 59. FINAL TEST METRICS TABLE
# ============================================================

final_metrics = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Balanced Accuracy",

        "Macro Precision",

        "Macro Recall",

        "Macro F1",

        "Weighted F1",

        "ROC-AUC",

        "PR-AUC",

        "MCC",

        "Binary Attack Recall",

        "Binary Attack F1",

        "Binary Attack FNR"

    ],

    "Independent_Test": [

        test_accuracy,

        test_balanced_accuracy,

        test_macro_precision,

        test_macro_recall,

        test_macro_f1,

        test_weighted_f1,

        test_auc,

        test_pr_auc,

        test_mcc,

        test_binary_recall,

        test_binary_f1,

        test_binary_fnr

    ]

})


display(
    final_metrics.style.format(
        {
            "Independent_Test":
                "{:.4f}"
        }
    )
)


final_metrics.to_csv(

    os.path.join(
        OUTPUT_DIR,
        "Final_Independent_Test_Metrics.csv"
    ),

    index=False

)


# ============================================================
# 60. SAVE FINAL PREPROCESSING ARTIFACTS
# ============================================================

preprocessing_artifacts = {

    "target_column":
        TARGET_COLUMN,

    "target_encoder":
        target_encoder,

    "categorical_encoder":
        encoder,

    "categorical_columns":
        categorical_columns,

    "numeric_columns":
        numeric_columns,

    "train_medians":
        complete_medians,

    "variance_selector":
        final_variance_selector,

    "variance_features":
        final_variance_features,

    "scaler":
        final_scaler,

    "correlation_threshold":
        CORRELATION_THRESHOLD,

    "correlated_features":
        final_correlated_features,

    "selected_features":
        final_selected_features,

    "class_names":
        CLASS_NAMES,

    "class_weights":
        complete_class_weights,

    "ensemble_weights":
        ensemble_weights

}


joblib.dump(

    preprocessing_artifacts,

    os.path.join(
        OUTPUT_DIR,
        "NSL_KDD_Final_Preprocessing.joblib"
    )

)


# ============================================================
# 61. SAVE FINAL MODELS
# ============================================================

model_directory = os.path.join(

    OUTPUT_DIR,

    "Final_Models"

)


os.makedirs(

    model_directory,

    exist_ok=True

)


for name, model in final_models.items():

    safe_name = (

        name
        .replace(
            " ",
            "_"
        )

    )


    joblib.dump(

        model,

        os.path.join(

            model_directory,

            safe_name + ".joblib"

        )

    )


# ============================================================
# 62. SAVE EXPERIMENT CONFIGURATION
# ============================================================

configuration = {

    "random_seed":
        SEED,

    "training_samples":
        len(train_df),

    "independent_test_samples":
        len(test_df),

    "validation_fraction":
        VALIDATION_SIZE,

    "correlation_threshold":
        CORRELATION_THRESHOLD,

    "top_k_features":
        TOP_K_FEATURES,

    "cv_folds":
        N_CV_FOLDS,

    "target":
        TARGET_COLUMN,

    "classes":
        CLASS_NAMES,

    "selected_features":
        final_selected_features,

    "ensemble_weights":
        ensemble_weights,

    "best_model":
        best_model_name

}


joblib.dump(

    configuration,

    os.path.join(

        OUTPUT_DIR,

        "Experiment_Configuration.joblib"

    )

)


# ============================================================
# 63. DEPLOYMENT FUNCTION
# ============================================================

def predict_network_intrusion(

    new_data,

    return_probabilities=True

):

    """
    Predict NSL-KDD intrusion classes using the final
    hybrid ensemble.

    Parameters
    ----------
    new_data : pandas.DataFrame
        Raw network traffic observations.

    return_probabilities : bool
        Whether to return class probabilities.

    Returns
    -------
    pandas.DataFrame
        Predicted class, confidence and optionally
        class probabilities.
    """

    data = new_data.copy()


    # --------------------------------------------------------
    # Remove target if present
    # --------------------------------------------------------

    if TARGET_COLUMN in data.columns:

        data = data.drop(

            columns=[
                TARGET_COLUMN
            ]

        )


    # --------------------------------------------------------
    # Remove index columns
    # --------------------------------------------------------

    data = remove_index_columns(
        data
    )


    # --------------------------------------------------------
    # Align columns
    # --------------------------------------------------------

    expected_columns = (

        X_complete_raw.columns
    )


    data = data.reindex(

        columns=expected_columns

    )


    # --------------------------------------------------------
    # Encode categorical variables
    # --------------------------------------------------------

    if categorical_columns:

        data[
            categorical_columns
        ] = encoder.transform(

            data[
                categorical_columns
            ].astype(str)

        )


    # --------------------------------------------------------
    # Numeric conversion
    # --------------------------------------------------------

    for column in data.columns:

        data[column] = pd.to_numeric(

            data[column],

            errors="coerce"

        )


    # --------------------------------------------------------
    # Imputation
    # --------------------------------------------------------

    data = data.fillna(

        complete_medians

    )


    # --------------------------------------------------------
    # Variance
    # --------------------------------------------------------

    data_var = (

        final_variance_selector
        .transform(data)

    )


    data_var = pd.DataFrame(

        data_var,

        columns=final_variance_features

    )


    # --------------------------------------------------------
    # Scaling
    # --------------------------------------------------------

    data_scaled = (

        final_scaler
        .transform(data_var)

    )


    data_scaled = pd.DataFrame(

        data_scaled,

        columns=final_variance_features

    )


    # --------------------------------------------------------
    # Correlation
    # --------------------------------------------------------

    data_corr = (

        data_scaled
        .drop(
            columns=final_correlated_features,
            errors="ignore"
        )

    )


    # --------------------------------------------------------
    # Feature selection
    # --------------------------------------------------------

    data_selected = (

        data_corr[
            final_selected_features
        ]

    )


    # --------------------------------------------------------
    # Ensemble prediction
    # --------------------------------------------------------

    ensemble_probability = np.zeros(

        (
            len(data_selected),
            NUM_CLASSES
        )

    )


    for name, model in final_models.items():

        probability = model.predict_proba(

            data_selected

        )


        weight = (

            ensemble_weights
            .get(
                name,
                0
            )

        )


        ensemble_probability += (

            weight *
            probability

        )


    prediction = np.argmax(

        ensemble_probability,

        axis=1

    )


    predicted_labels = (

        target_encoder
        .inverse_transform(
            prediction
        )

    )


    confidence = (

        np.max(
            ensemble_probability,
            axis=1
        )

    )


    output = pd.DataFrame({

        "Predicted_Class":
            predicted_labels,

        "Confidence":
            confidence

    })


    if return_probabilities:

        for class_id, class_name in enumerate(
            CLASS_NAMES
        ):

            output[
                f"Probability_{class_name}"
            ] = (

                ensemble_probability[
                    :,
                    class_id
                ]

            )


    return output


# ============================================================
# 64. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 110)
print("=" * 110)


print(
    "\nTraining samples:",
    len(train_df)
)


print(
    "Independent test samples:",
    len(test_df)
)


print(
    "Number of classes:",
    NUM_CLASSES
)


print(
    "Final selected features:",
    len(final_selected_features)
)


print(
    "\nClasses:"
)


print(
    CLASS_NAMES
)


print(
    "\nFinal hybrid ensemble:"
)


for name, weight in ensemble_weights.items():

    print(
        f"  {name}: {weight:.4f}"
    )


print(
    "\nIndependent test Macro-F1:",
    f"{test_macro_f1:.4f}"
)


print(
    "Independent test Balanced Accuracy:",
    f"{test_balanced_accuracy:.4f}"
)


print(
    "Independent test ROC-AUC:",
    f"{test_auc:.4f}"
)


print(
    "\nOutput directory:"
)


print(
    OUTPUT_DIR
)


print("\nGenerated outputs include:")

print(
    "1. Class_Distribution.csv"
)

print(
    "2. Five_Fold_Cross_Validation.csv"
)

print(
    "3. Complete_Validation_Model_Comparison.csv"
)

print(
    "4. Final_Independent_Test_Metrics.csv"
)

print(
    "5. Final_Test_Classification_Report.txt"
)

print(
    "6. Validation_False_Negative_Analysis.csv"
)

print(
    "7. Final_Test_False_Negative_Analysis.csv"
)

print(
    "8. Final_Test_Confusion_Matrix.png"
)

print(
    "9. Final_Test_Normalized_Confusion_Matrix.png"
)

print(
    "10. Final_Test_ROC_Curves.png"
)

print(
    "11. Final_Test_PR_Curves.png"
)

print(
    "12. Selected_Feature_Importance.png"
)

print(
    "13. Final_XGBoost_Feature_Importance.png"
)

print(
    "14. NSL_KDD_Final_Preprocessing.joblib"
)

print(
    "15. Final_Models/"
)

print(
    "16. Experiment_Configuration.joblib"
)

print(
    "17. Model_Comparison.png"
)

print(
    "\n✓ Independent test set was not used for feature selection."
)

print(
    "✓ Independent test set was not used for model selection."
)

print(
    "✓ Feature preprocessing was learned from training data."
)

print(
    "✓ Class-wise performance and false negatives were evaluated."
)

print(
    "✓ Five-fold cross-validation was performed."
)

print(
    "✓ Hybrid ensemble evaluation was included."
)
```
